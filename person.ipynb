```python
import matplotlib.pyplot as plt
import pandas
import pandas as pd
from scipy.stats import pearsonr, chi2_contingency
from sklearn.feature_selection import mutual_info_classif
import pandas as pd
# بيانات بسيطة
data = pd.DataFrame({
    'Age': [20, 22, 25, 30, 35, 40, 45, 50],
    'Salary': [2000, 2200, 2500, 3000, 3500, 4000, 4500, 5000],
    'Purchased': [0, 0, 0, 1, 1, 1, 1, 1]
})

print(data)
```

```python

# -------------------------
# 1- Pearson Correlation
# -------------------------
corr, p_value = pearsonr(data['Age'], data['Salary'])

print("Pearson Correlation:")
print("Correlation =", round(corr, 4))
print("P-value =", round(p_value, 4))
```

```python

# -------------------------
# 2- Chi-Square
# -------------------------
data['AgeGroup'] = ['Young', 'Young', 'Young', 'Adult',
                    'Adult', 'Adult', 'Senior', 'Senior']

contingency_table = pd.crosstab(data['AgeGroup'], data['Purchased'])

chi2, p, dof, expected = chi2_contingency(contingency_table)

print("\nChi-Square:")
print("Chi2 =", round(chi2, 4))
print("P-value =", round(p, 4))
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder, MinMaxScaler
from sklearn.tree import DecisionTreeClassifier

# البيانات
data = {
    "City": ["Miami", "New York", "Orlando", "Miami", "Phoenix"],
    "Age": [42, 32, 18, 45, 35],
    "Label": ["yes", "no", "no", "yes", "no"]
}

df = pd.DataFrame(data)
print(df)
```

```python

# تحويل البيانات النصية إلى أرقام
city_encoder = LabelEncoder()
label_encoder = LabelEncoder()

df["City"] = city_encoder.fit_transform(df["City"])
df["Label"] = label_encoder.fit_transform(df["Label"])

# الميزات والهدف
X = df[["City", "Age"]]
y = df["Label"]
```

```python

# بناء النموذج وتدريبه
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)
```

```python

# بيانات الاختبار من الجدول
test_data = pd.DataFrame({
    "City": ["Chicago", "New York"],
    "Age": [72, 48]
})

# تحويل المدينة إلى أرقام
all_cities = ["Miami", "New York", "Orlando", "Phoenix", "Chicago"]
city_encoder.fit(all_cities)

test_data["City"] = city_encoder.transform(test_data["City"])
```

```python

# التنبؤ
predictions = model.predict(test_data)
```

```python

# تحويل النتائج إلى yes/no
predictions = label_encoder.inverse_transform(predictions)

print(predictions)
```
